```python
# -*- coding: utf-8 -*-
"""untitled20.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1cy505WtpI5k4Zg3hIDxvyJ_Bf-3qz6OR

### Bitcoin address balance inspector

Read-only balance lookups against the public Blockchain.com API.

What changed in this pass (the point is to surface *palpable* hits, not noise):

* a "hit" now means **funded** (``balance > 0``), previously it was any address
  used on-chain at least once -- which flagged almost every address and was
  therefore useless for finding real value;
* hits are ranked by descending balance and tagged with a value tier
  (dust / micro / small / moderate / notable / large / whale);
* results stream to disk as they are discovered, so memory is bounded by the
  number of *hits* rather than the number of addresses scanned -- arbitrarily
  large lists no longer blow up RAM and hits survive a crash;
* ``--min-btc`` / ``--top`` let you ask only for hit-worthy results;
* aggregate summary reports how much value was actually found;
* the correctness guarantees from before are kept: retries + exponential
  backoff, request chunking, recursive chunk-split fallback, and a failed
  lookup is reported as UNKNOWN rather than silently treated as 0 (a swallowed
  error would hide real hits);
* no accidental module-level side effects -- everything runs behind ``main()``.

Public API:
    BlockchainInfoClient   -- the client
    AddressBalance         -- one immutable result row
    get_btc_balance(...)   -- one-shot convenience helper
"""

from __future__ import annotations

import argparse
import csv
import json
import logging
import sys
import time
from collections import Counter
from concurrent.futures import ThreadPoolExecutor
from dataclasses import asdict, dataclass, field
from typing import Dict, Iterable, Iterator, List, Optional, Sequence

import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

SATOSHIS_PER_BTC = 100_000_000
BLOCKCHAIN_INFO_URL = "https://blockchain.info/balance"

# Blockchain.com tolerates reasonably long ids, but staying well below typical
# URL/proxy limits keeps every request safe.
DEFAULT_CHUNK_SIZE = 80
DEFAULT_WORKERS = 8
DEFAULT_TIMEOUT = 20.0
DEFAULT_RETRIES = 5
DEFAULT_BACKOFF = 0.7

DEFAULT_ADDRESSES = (
    "1JTK7s9YVYywfm5XUH7RNhHJH1LshCaRFR",
    "1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNa",  # Bitcoin genesis coinbase address
)

# Value tiers, highest floor first. A "palpable" hit is anything at or above
# ``PALPABLE_MIN_SAT`` (default: "small"); dust is technically funded but not
# worth flagging as a hit.
DUST_SAT = 546  # the classic P2PKH dust threshold
PALPABLE_MIN_SAT = 100_000  # 0.001 BTC
TIER_FLOORS = (
    ("whale", 100 * SATOSHIS_PER_BTC),
    ("large", 10 * SATOSHIS_PER_BTC),
    ("notable", 1 * SATOSHIS_PER_BTC),
    ("moderate", SATOSHIS_PER_BTC // 10),  # 0.1 BTC
    ("small", PALPABLE_MIN_SAT),  # 0.001 BTC
)


def _tier_for(balance_sat: Optional[int]) -> str:
    """Classify a balance so hits can be triaged at a glance."""
    if balance_sat is None:
        return "unknown"
    if balance_sat <= 0:
        return "empty"
    for name, floor in TIER_FLOORS:
        if balance_sat >= floor:
            return name
    return "dust" if balance_sat <= DUST_SAT else "micro"


# --------------------------------------------------------------------------- #
# Result model
# --------------------------------------------------------------------------- #
@dataclass(frozen=True)
class AddressBalance:
    """Immutable result for a single address.

    ``balance_sat``/``n_tx`` are ``None`` when the lookup failed, so callers can
    distinguish "confirmed empty" (0) from "we do not know" (``None``).
    """

    address: str
    balance_sat: Optional[int] = None
    n_tx: Optional[int] = None
    error: Optional[str] = None

    @property
    def known(self) -> bool:
        return self.error is None and self.balance_sat is not None

    @property
    def balance_btc(self) -> Optional[float]:
        if self.balance_sat is None:
            return None
        return self.balance_sat / SATOSHIS_PER_BTC

    @property
    def is_used(self) -> bool:
        """Ever touched on-chain (may still be drained/empty)."""
        return bool(self.n_tx)

    @property
    def is_funded(self) -> bool:
        """Holds a non-zero balance right now."""
        return bool(self.balance_sat)

    @property
    def is_hit(self) -> bool:
        """A *palpable* hit: an address that actually holds value."""
        return self.is_funded

    @property
    def tier(self) -> str:
        return _tier_for(self.balance_sat)

    def to_row(self) -> Dict[str, object]:
        row = asdict(self)
        row["balance_btc"] = self.balance_btc
        row["tier"] = self.tier
        return row


# --------------------------------------------------------------------------- #
# Client
# --------------------------------------------------------------------------- #
class BlockchainInfoClient:
    """Concurrent, retrying, chunking client for ``blockchain.info/balance``."""

    def __init__(
        self,
        *,
        session: Optional[requests.Session] = None,
        timeout: float = DEFAULT_TIMEOUT,
        retries: int = DEFAULT_RETRIES,
        backoff: float = DEFAULT_BACKOFF,
        chunk_size: int = DEFAULT_CHUNK_SIZE,
    ) -> None:
        if chunk_size < 1:
            raise ValueError("chunk_size must be >= 1")
        self._timeout = timeout
        self._chunk_size = chunk_size
        self._session = session or self._build_session(retries, backoff)

    # -- session plumbing -------------------------------------------------- #
    @staticmethod
    def _build_session(retries: int, backoff: float) -> requests.Session:
        retry = Retry(
            total=retries,
            connect=retries,
            read=retries,
            status=retries,
            backoff_factor=backoff,
            status_forcelist=(429, 500, 502, 503, 504),
            allowed_methods=frozenset({"GET"}),
            respect_retry_after_header=True,
            raise_on_status=False,
        )
        adapter = HTTPAdapter(max_retries=retry, pool_connections=32, pool_maxsize=32)
        session = requests.Session()
        session.mount("https://", adapter)
        session.mount("http://", adapter)
        session.headers.update({"User-Agent": "btc-balance-inspector/3.0"})
        return session

    def close(self) -> None:
        self._session.close()

    def __enter__(self) -> "BlockchainInfoClient":
        return self

    def __exit__(self, *exc: object) -> None:
        self.close()

    # -- transport --------------------------------------------------------- #
    def _fetch_chunk(self, addresses: Sequence[str]) -> Dict[str, dict]:
        response = self._session.get(
            BLOCKCHAIN_INFO_URL,
            params={"active": ",".join(addresses)},
            timeout=self._timeout,
        )
        response.raise_for_status()
        payload = response.json()
        if not isinstance(payload, dict):
            raise ValueError(f"unexpected payload type: {type(payload).__name__}")
        return payload

    def _fetch_with_fallback(
        self, chunk: Sequence[str]
    ) -> tuple[Dict[str, dict], Optional[str]]:
        """Fetch a chunk; on failure split it and retry, recursing to single
        addresses so a single bad id can never sink a whole batch."""
        try:
            return self._fetch_chunk(chunk), None
        except Exception as exc:  # noqa: BLE001 - surfaced via the error field
            if len(chunk) > 1:
                mid = len(chunk) // 2
                left, left_err = self._fetch_with_fallback(chunk[:mid])
                right, right_err = self._fetch_with_fallback(chunk[mid:])
                left.update(right)
                return left, left_err or right_err
            return {}, f"{type(exc).__name__}: {exc}"

    # -- public lookup API ------------------------------------------------- #
    def iter_balances(
        self, addresses: Iterable[str], *, workers: int = DEFAULT_WORKERS
    ) -> Iterator[AddressBalance]:
        """Yield results lazily. Memory stays bounded by
        ``chunk_size * workers`` no matter how many addresses are supplied."""
        workers = max(1, workers)
        batch: List[str] = []
        threshold = self._chunk_size * workers
        for address in addresses:
            batch.append(address)
            if len(batch) >= threshold:
                yield from self._process(batch, workers)
                batch = []
        if batch:
            yield from self._process(batch, workers)

    def _process(self, addresses: Sequence[str], workers: int) -> Iterator[AddressBalance]:
        unique = list(dict.fromkeys(addresses))  # de-dupe, preserve order
        chunks = list(_chunked(unique, self._chunk_size))
        with ThreadPoolExecutor(max_workers=min(workers, len(chunks))) as pool:
            # pool.map preserves chunk order, so zip stays aligned.
            for chunk, (data, err) in zip(chunks, pool.map(self._fetch_with_fallback, chunks)):
                for address in chunk:
                    info = data.get(address)
                    if info is None:
                        if err is not None:
                            yield AddressBalance(address, error=err)
                        else:
                            yield AddressBalance(address, 0, 0)
                        continue
                    sats = int(info.get("final_balance", 0) or 0)
                    n_tx = int(info.get("n_tx", 0) or 0)
                    yield AddressBalance(address, sats, n_tx)

    def get_many(
        self, addresses: Iterable[str], *, workers: int = DEFAULT_WORKERS
    ) -> Dict[str, AddressBalance]:
        return {bal.address: bal for bal in self.iter_balances(addresses, workers=workers)}

    def get(self, address: str, *, workers: int = 1) -> Optional[AddressBalance]:
        return next(iter(self.iter_balances([address], workers=workers)), None)


# --------------------------------------------------------------------------- #
# Helpers
# --------------------------------------------------------------------------- #
def _chunked(items: Sequence[str], size: int) -> Iterator[Sequence[str]]:
    for start in range(0, len(items), size):
        yield items[start : start + size]


def get_btc_balance(
    addresses, *, workers: int = DEFAULT_WORKERS, **client_kwargs
) -> Dict[str, AddressBalance]:
    """Convenience one-shot helper.

    Accepts a single address (returns that ``AddressBalance``) or an iterable
    (returns a mapping of address -> ``AddressBalance``).
    """
    single = isinstance(addresses, str)
    address_list = [addresses] if single else list(addresses)
    with BlockchainInfoClient(**client_kwargs) as client:
        results = client.get_many(address_list, workers=workers)
    return results[addresses] if single else results


def rank_hits(
    balances: Iterable[AddressBalance], *, min_sat: int = 1
) -> List[AddressBalance]:
    """Return funded addresses at or above ``min_sat``, richest first."""
    hits = [
        bal for bal in balances if bal.is_funded and (bal.balance_sat or 0) >= min_sat
    ]
    hits.sort(key=lambda bal: bal.balance_sat or 0, reverse=True)
    return hits


@dataclass
class ScanSummary:
    """Aggregate outcome of a scan."""

    min_sat: int = 1
    checked: int = 0
    funded: int = 0        # palpable hits (>= min_sat)
    funded_below_min: int = 0  # funded but under the palpable threshold
    used_only: int = 0     # ever used on-chain but currently empty
    empty: int = 0
    unknown: int = 0
    total_sat: int = 0
    tiers: Counter = field(default_factory=Counter)

    @property
    def total_btc(self) -> float:
        return self.total_sat / SATOSHIS_PER_BTC

    def observe(self, bal: AddressBalance) -> None:
        self.checked += 1
        if not bal.known:
            self.unknown += 1
            return
        if bal.is_funded:
            if (bal.balance_sat or 0) >= self.min_sat:
                self.funded += 1
                self.total_sat += bal.balance_sat or 0
                self.tiers[bal.tier] += 1
            else:
                self.funded_below_min += 1
        elif bal.is_used:
            self.used_only += 1
        else:
            self.empty += 1


def print_hits(hits: Sequence[AddressBalance], *, top: Optional[int] = None) -> None:
    shown = hits if top is None else hits[:top]
    for rank, bal in enumerate(shown, 1):
        print(
            f"#{rank:<4} {bal.address}\t{bal.balance_btc:.8f} BTC\t"
            f"({bal.balance_sat} sat)\t{bal.n_tx} tx\t[{bal.tier}]"
        )
    if top is not None and len(hits) > top:
        print(f"... and {len(hits) - top} more hit(s) below the cut.")


def print_summary(summary: ScanSummary, elapsed: float) -> None:
    print(
        f"\n{summary.checked} address(es) checked in {elapsed:.2f}s | "
        f"{summary.funded} funded hit(s) | {summary.funded_below_min} below threshold | "
        f"{summary.used_only} used-but-empty | {summary.empty} empty | "
        f"{summary.unknown} unknown"
    )
    if summary.funded:
        breakdown = ", ".join(f"{name}: {count}" for name, count in summary.tiers.most_common())
        print(f"Recovered value: {summary.total_btc:.8f} BTC  ({breakdown})")


# --------------------------------------------------------------------------- #
# Streaming sinks
# --------------------------------------------------------------------------- #
class ResultStream:
    """Writes hits to disk as they are found.

    Keeping only hits in memory means a scan of millions of addresses costs the
    same as a scan of a handful, and every hit is durable the moment it is seen.
    """

    def __init__(
        self,
        *,
        csv_path: Optional[str] = None,
        jsonl_path: Optional[str] = None,
        min_sat: int = 1,
    ) -> None:
        self.min_sat = min_sat
        self._csv = open(csv_path, "w", newline="", encoding="utf-8") if csv_path else None
        self._jsonl = open(jsonl_path, "a", encoding="utf-8") if jsonl_path else None
        self._writer: Optional[csv.DictWriter] = None
        self.csv_path = csv_path
        self.jsonl_path = jsonl_path
        self.written = 0

    def __enter__(self) -> "ResultStream":
        return self

    def __exit__(self, *exc: object) -> None:
        self.close()

    def close(self) -> None:
        for handle in (self._csv, self._jsonl):
            if handle is not None and not handle.closed:
                handle.close()

    def offer(self, bal: AddressBalance) -> bool:
        if not (bal.is_funded and (bal.balance_sat or 0) >= self.min_sat):
            return False
        row = bal.to_row()
        if self._jsonl is not None:
            self._jsonl.write(json.dumps(row, ensure_ascii=False) + "\n")
            self._jsonl.flush()
        if self._csv is not None:
            if self._writer is None:
                self._writer = csv.DictWriter(self._csv, fieldnames=list(row))
                self._writer.writeheader()
            self._writer.writerow(row)
            self._csv.flush()
        self.written += 1
        return True


# --------------------------------------------------------------------------- #
# Offline export helpers (materialise an already-collected iterable)
# --------------------------------------------------------------------------- #
def to_dataframe(balances: Iterable[AddressBalance], *, hits_only: bool = False):
    """Build a pandas DataFrame (pandas is imported lazily)."""
    import pandas as pd

    rows = [bal.to_row() for bal in balances if not hits_only or bal.is_hit]
    columns = ["address", "balance_btc", "balance_sat", "n_tx", "tier", "error"]
    return pd.DataFrame(rows, columns=columns)


def export_csv(
    balances: Iterable[AddressBalance], path: str, *, hits_only: bool = False
) -> int:
    frame = to_dataframe(balances, hits_only=hits_only)
    frame.to_csv(path, index=False)
    return len(frame)


def export_json(
    balances: Iterable[AddressBalance], path: str, *, hits_only: bool = False
) -> int:
    rows = [bal.to_row() for bal in balances if not hits_only or bal.is_hit]
    with open(path, "w", encoding="utf-8") as handle:
        json.dump(rows, handle, indent=2)
    return len(rows)


def _read_addresses_file(path: str) -> Iterator[str]:
    with open(path, "r", encoding="utf-8") as handle:
        for raw in handle:
            line = raw.split("#", 1)[0].strip()
            if line:
                yield line


def _address_source(args: argparse.Namespace) -> Iterator[str]:
    yield from args.addresses
    for path in args.file or []:
        yield from _read_addresses_file(path)
    if args.stdin:
        for raw in sys.stdin:
            line = raw.split("#", 1)[0].strip()
            if line:
                yield line


# --------------------------------------------------------------------------- #
# CLI
# --------------------------------------------------------------------------- #
def _build_parser() -> argparse.ArgumentParser:
    parser = argparse.ArgumentParser(description="Check Bitcoin address balances.")
    parser.add_argument("addresses", nargs="*", help="addresses (defaults to the built-in list)")
    parser.add_argument(
        "-f", "--file", action="append", help="file with one address per line "
        "(# comments allowed); repeatable"
    )
    parser.add_argument("--stdin", action="store_true", help="also read addresses from stdin")
    parser.add_argument("--csv", help="write ranked hits to this CSV path")
    parser.add_argument("--jsonl", dest="jsonl_path", help="append hits as JSON lines (streamed)")
    parser.add_argument("--json", dest="json_path", help="write hits to this JSON path (one-shot)")
    parser.add_argument(
        "--min-btc", type=float, default=0.0,
        help="only count/report addresses holding at least this much (palpable threshold)",
    )
    parser.add_argument("--top", type=int, default=None, help="show only the N richest hits")
    parser.add_argument(
        "--hits-only", action="store_true", help="print funded hits only (no summary noise)"
    )
    parser.add_argument("--workers", type=int, default=DEFAULT_WORKERS)
    parser.add_argument("--chunk-size", type=int, default=DEFAULT_CHUNK_SIZE)
    parser.add_argument("--timeout", type=float, default=DEFAULT_TIMEOUT)
    parser.add_argument("-v", "--verbose", action="store_true")
    return parser


def main(argv: Optional[Sequence[str]] = None) -> int:
    args = _build_parser().parse_args(argv)
    logging.basicConfig(
        level=logging.DEBUG if args.verbose else logging.WARNING,
        format="%(asctime)s %(levelname)s %(name)s: %(message)s",
    )

    min_sat = max(0, int(round(args.min_btc * SATOSHIS_PER_BTC)))

    sources = _address_source(args)
    try:
        first = next(sources)
    except StopIteration:
        first = None
    addresses: Iterable[str] = sources if first is None else _chain_one(first, sources)
    if first is None:
        addresses = DEFAULT_ADDRESSES

    summary = ScanSummary(min_sat=min_sat)
    hits: List[AddressBalance] = []
    started = time.monotonic()

    with ResultStream(csv_path=args.csv, jsonl_path=args.jsonl_path, min_sat=min_sat) as stream:
        with BlockchainInfoClient(timeout=args.timeout, chunk_size=args.chunk_size) as client:
            # Consume lazily: only hits are retained, so RAM is independent of
            # how many addresses are scanned.
            for bal in client.iter_balances(addresses, workers=args.workers):
                summary.observe(bal)
                if stream.offer(bal):
                    hits.append(bal)

    if summary.checked == 0:
        print("No addresses to check.")
        return 1

    hits.sort(key=lambda bal: bal.balance_sat or 0, reverse=True)
    print_hits(hits, top=args.top)
    if not args.hits_only:
        print_summary(summary, time.monotonic() - started)

    if args.csv:
        print(f"Wrote {stream.written} hit(s) to {args.csv}")
    if args.jsonl_path:
        print(f"Appended {stream.written} hit(s) to {args.jsonl_path}")
    if args.json_path:
        count = export_json(hits, args.json_path)
        print(f"Wrote {count} hit(s) to {args.json_path}")

    return 0 if summary.unknown == 0 else 2


def _chain_one(first: str, rest: Iterable[str]) -> Iterator[str]:
    yield first
    yield from rest


if __name__ == "__main__":
    sys.exit(main())
```
